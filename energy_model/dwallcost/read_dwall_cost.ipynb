```python
import numpy as np
import matplotlib.pyplot as plt

def read_domain_wall_cost(filename):
    
    dwall_costs = []
    cost_change = []
    file  = open(filename)
    lines = file.readlines()
    for line in lines:
        if line[0]=='E':
            dwall_costs.append(float(line.split(':')[1]))
        if line[0]=='P':
            cost_change.append(float(line.split('=')[1]))
    return dwall_costs,cost_change
```

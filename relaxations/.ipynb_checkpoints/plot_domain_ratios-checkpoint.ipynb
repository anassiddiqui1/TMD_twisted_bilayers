```python
import math
import numpy as np
import matplotlib.pyplot as plt
from ase.io import read,write;
from ase import Atoms;
from ase.build.surface import mx2
from os import environ;
# from ase.io.lammpsdata import write_lammps_data, read_lammps_data;
import sys
from glob import glob
from ase.visualize import view
from curses import meta
import math;
import numpy as np;
import matplotlib.pyplot as plt;
import matplotlib as mpl;
from ase import Atoms;
from ase.io import write, read;
from scipy.interpolate import griddata
from mpl_toolkits.axes_grid1 import make_axes_locatable
import os
from scipy.spatial.distance import cdist
from scipy.integrate import quad
from utils import *
import matplotlib.gridspec as gridspec



def get_continuous_curve(xdata,ydata,max_distance):
    
    k=True
    lower,higher = 0,len(xdata)
    #prev_distance = calculate_distance([xdata[1],ydata[1]],[xdata[0],ydata[0]])
    #prev_distance = abs(ydata[1]-ydata[0])
    for i in range(1,len(xdata)):
        #distance  = calculate_distance([xdata[i],ydata[i]],[xdata[i-1],ydata[i-1]])
        distance = abs(ydata[i]-ydata[i-1])
        #if distance>2*prev_distance:
        if distance>max_distance:
            print(distance,prev_distance)
            if k:
                lower = i
                k=False
            else:
                higher = i-1
                break
        else:
            prev_distance = distance
        
    return lower,higher

def get_max_ds(x_sc,y_sc,d_local_sc,x_min,x_max,y_min,y_max,bool_x,spacing=100):
    
    xdata = []
    ydata = []
    ddata = []

    xspace = np.linspace(x_min,x_max,spacing)
    yspace = np.linspace(y_min,y_max,spacing)
    xs,ys = np.meshgrid(xspace,yspace)
    ds = griddata(points=(x_sc, y_sc), values=d_local_sc, xi=(xs, ys), method=GRIDDATA_ORDER)
    
    increasing = False
    if bool_x:
        for i,x in enumerate(xspace):
            max_arg = np.argmax(ds[:,i])
            ddata.append(ds[max_arg][i])
            xdata.append(xs[max_arg][i])
            ydata.append(ys[max_arg][i])
            
#             if i>0 and ydata[i]>=ydata[i-1]:
#                 increasing=True
            
#             if increasing and ydata[i]<ydata[i-1]:
#                 print('Maxima exists')
                
            
    else:
        for i,x in enumerate(xspace):
            max_arg = np.argmax(ds[i,:])
            xdata.append(xs[i][max_arg])
            ydata.append(ys[i][max_arg])
            ddata.append(ds[i][max_arg])
    #print(ddata)
    return xdata,ydata


def slope_intercept_line(slope_degrees, point):
    # Convert degrees to radians
    slope_radians = math.radians(slope_degrees)
    
    # Slope of the line
    m = math.tan(slope_radians)
    
    # Point (x1, y1)
    x1, y1 = point
    
    # Use the point-slope form: y - y1 = m(x - x1)
    # Convert it to slope-intercept form: y = mx + (y1 - mx1)
    b = y1 - m * x1
    
    return m,b
```

```python
def MakeGrid(cell, theta, A_ML):
    if theta>30: theta = 60 - theta # transform to 60°-theta
    a_moire = A_ML / np.sqrt(2*(1-np.cos(np.deg2rad(theta)))) # moiré cell constant approximated by monolayer (does not include compression in tBL)
    ## identify a suitable plotting reference depending on moiré cell size (from trial and error)
#     if a_moire<=20: A_REF = 20 # tiny cell size
#     elif a_moire>20 and a_moire<=50: A_REF = 50 # small cell size
#     elif a_moire>50 and a_moire<=250: A_REF = 200 # intermediate size
#     #elif a_moire>50 and a_moire<=200: A_REF = 500 # intermediate size
#     else: A_REF = 1000 # large >200 Ang
    if a_moire<=23: A_REF = 20 # tiny cell size
    elif a_moire>23 and a_moire<=65: A_REF = 50 # small cell size
    elif a_moire>65 and a_moire<=250: A_REF = 200 # intermediate size
    #elif a_moire>50 and a_moire<=200: A_REF = 500 # intermediate size
    else: A_REF = 1000 # large >200 Ang
        
    #print(f'A_REF={A_REF} a_moire={a_moire}')
    ## set up grid steps from cell size
    a = np.sqrt(cell[0][0]**2 + cell[0][1]**2 + cell[0][2]**2) # cell constant
    GRID_STEPS = int(77.832*(a/A_REF)+9.8418)+1
    
    #GRID_STEPS = int(77.832*(a/A_REF)+9.8418)+1 # linear function fitted to results from previous tests what looks nice without artifacts
#     if A_REF==20: GRID_STEPS = int(77.832*(a/A_REF)+9.8418)+1
#     elif A_REF==50: GRID_STEPS = int(77.832*(a/A_REF)+9.8418)+1
#     if A_REF==200: GRID_STEPS = int(77.832*(a/A_REF)+9.8418)+1
#     else: GRID_STEPS = int(77.832*(a/A_REF)+9.8418)+1
    print(a_moire,A_REF)
    ## create grid
    x_grid, y_grid = [], [] # new lists
    for i in range(0, GRID_STEPS): # along a direction
        for k in range(0, GRID_STEPS): # along b direction
            x_grid.append(i*cell[0][0]/GRID_STEPS + k*cell[1][0]/GRID_STEPS)
            y_grid.append(i*cell[0][1]/GRID_STEPS + k*cell[1][1]/GRID_STEPS)
    ## return statement
    return x_grid, y_grid, A_REF
```

```python
def find_max_d_on_line(x_sc, y_sc, d_local_sc, x_i, y1, y2):
    # Filter points where x == x_i
    x_condition = (x_sc.round(5) == x_i)
    
    # Further filter points that are within the specified y range
    y_condition = (y_sc.round(5) >= y1) & (y_sc.round(5) <= y2)
    
    # Combine both conditions
    combined_condition = x_condition & y_condition
    
    
    # Apply the combined condition to the d_local_sc array to get the relevant d values
    relevant_d_indices = np.where(combined_condition)[0]
    #print(len(relevant_d_indices),np.argmax(d_local_sc[relevant_d_indices]))
    sc_index = relevant_d_indices[np.argmax(d_local_sc[relevant_d_indices])]
    return sc_index

def find_max_d_points(x_sc, y_sc, d_local_sc,x1, x2, y1, y2):
    x_s = np.unique(x_sc[(x_sc>=x_sol2) & (x_sc<=x_sol3)].round(5))
    indices = np.zeros(len(x_s),dtype=int)
    for i,x in enumerate(x_s):
        indices[i] = find_max_d_on_line(x_sc, y_sc, d_local_sc, x, y1, y2)  
    #return x_sc[indices[::2]],y_sc[indices[::2]]
    return x_sc[indices],y_sc[indices]
```

```python
### Works perfectly for homobilayers

from matplotlib.patches import Arc

colormap='YlGnBu_r'
stacking = 'AP'
poly_deg = 5
spacing = 100
SIZE_PT = 4 
ratios_dict = {}

for stacking in ['P','AP']:
    if stacking =='P':
        mpairs = [('Mo','W')]
    else:
        mpairs = [('Mo','Mo'),('Mo','W')]
        
        
    for m1,m2 in mpairs:
#         if m1==m2:
#             iis=[10,12,15,16,18,20,22,25,27,30,40]
#         else:
#             iis=[10,12,15,16,18,20,22,25,27,30]
        for x1 in ['S','Se']:
            
            if m1==m2 or x1=='Se':
                iis=[10,12,15,16,18,20,22,25,27,30,40]
            else:
                iis=[10,12,15,16,18,20,22,25,27,30]
                
                
            ratios = np.zeros(len(iis))
            print(f'Starting calculations for {stacking}-{m1}{x1}2/{m2}{x1}2...')
            for k,ii in enumerate(iis):
                print(f'ii={ii}')
                A_ML = 0.5*(a_optb88[f'{m1}{x1}2']+a_optb88[f'{m2}{x1}2'])


                if m1!=m2 and ii>10 and x1=='S':
                    filename = f'data_files/{m1}{x1}2_{m2}{x1}2_{stacking}_{ii}_twirl.xyz'
                else:
                    filename = f'data_files/{m1}{x1}2_{m2}{x1}2_{stacking}_{ii}.xyz'

                image_file = f'ratio_figures/{m1}{x1}2_{m2}{x1}2_{stacking}_{ii}.png'
                # if os.path.exists(image_file):
                #     print(f'heatmaps/{m1}{x1}2_{m2}{x1}2_{stacking}_{ii}.png already exists.')
                #     continue
                # if not os.path.exists(filename):
                #     continue
                # cell = get_bilayer_cell(m1,m2,x1,ii,stacking)
                atoms = read(filename)
                cell = atoms.cell
                if stacking == 'P':
                    twist_angle = twist_angles[ii]
                elif stacking =='AP':
                    twist_angle = 60-twist_angles[ii]
                coord_LL,coord_UL = extract_positions(atoms)
                x_UL, y_UL, z_UL = coord_UL[:,0], coord_UL[:,1], coord_UL[:,2]
                x_LL, y_LL, z_LL = coord_LL[:,0], coord_LL[:,1], coord_LL[:,2]

                # extend coordinates to avoid interpolation artifacts
                x_UL_ext, y_UL_ext, z_UL_ext = ExtendCoordinates(x_UL, y_UL, z_UL, cell)
                x_LL_ext, y_LL_ext, z_LL_ext = ExtendCoordinates(x_LL, y_LL, z_LL, cell)
                # calculate interlayer distance
                x_grid, y_grid, A_REF = MakeGrid(cell, twist_angle,A_ML) # create interpolation grid
                z_UL_int = DoInterpolation(x_grid, y_grid, x_UL_ext, y_UL_ext, z_UL_ext) # interpolate UL
                z_LL_int = DoInterpolation(x_grid, y_grid, x_LL_ext, y_LL_ext, z_LL_ext) # interpolate LL
                d_local = GetInterlayerDistance(z_UL_int, z_LL_int) # calculate interlayer distance
                D_MIN, D_MAX = min(d_local), max(d_local) # Ang, interlayer distance references (min & max) to have consistent colorbar
                #print(d_local)
                ### perform plotting (not separated into functions)

                ## prepare parameters: size plotting area
                # window of (x,y) coord to plot relative to origin [Ang]
                X_NEG, X_POS = A_REF, 2*A_REF
                Y_NEG, Y_POS = int(0.5*A_REF), int(1.5*A_REF)
                # get cell constants
                a = np.sqrt(cell[0][0]**2+cell[0][1]**2+cell[0][2]**2) # cell constant a
                b = np.sqrt(cell[1][0]**2+cell[1][1]**2+cell[1][2]**2) # cell constant b
                # find how much to extend in terms of supercells
                x_left = math.ceil((X_NEG + 0.5*a)/a)+2
                x_right = math.ceil(X_POS/a)+1
                y_bottom = math.ceil(Y_NEG/(SIN60*b))+1
                y_top = math.ceil(Y_POS/(SIN60*b))+1

                ## prepare interpolated data: extend within supercell of plotting area
                x_sc, y_sc, d_local_sc = [], [], [] # prepare lists
                for m in range(-x_left-1, x_right+1): # supercell in x-direction
                    for n in range(-y_bottom-1, y_top+1): # supercell in y-direction
                        for i in range(len(d_local)): # over all data points
                            x_sc.append(x_grid[i]+m*cell[0][0]+n*cell[1][0]) # x coordinates
                            y_sc.append(y_grid[i]+m*cell[0][1]+n*cell[1][1]) # y coordinates
                            d_local_sc.append(d_local[i]) # local interlayer distance

                # initialize plot object
                fig = plt.figure() # initialize figure
                ax = fig.add_subplot(111, aspect='equal') # equal aspect for axes
                # decide parameters for line plotting
                width = 100
                linestyle = 'k-' 
                # get size of length bar depending on A_REF
                if A_REF==20: SIZE_BAR = 10 # 1 nm bar
                elif A_REF==50: SIZE_BAR = 20 # 2 nm bar
                elif A_REF==200: SIZE_BAR = 100 # 10 nm bar
                else: SIZE_BAR = 500 # 50 nm bar
                # get twist angle with correct number of decimals
                if twist_angle>=1 and twist_angle<=59: # two decimal places
                    angle_string = r"$\theta=$"+str("{:5.2f}".format(twist_angle) )+'°' # 2 characters before decimal, 2 after
                else: angle_string = r"$\theta=$"+str("{:5.3f}".format(twist_angle) )+'°' # 2 characters before decimal, 3 after
                # define position of length bar in bottom right
                OFFSET_X = 0.30*A_REF # position from right side
                OFFSET_Y = 0.20*A_REF # position from bottom
                WIDTH_Y =  0.04*A_REF # height of vertical lines
                # define position where twist angle is shown on top left
                LABEL_X = -X_NEG + 2.2/50*A_REF 
                LABEL_Y = Y_POS - 7.2/50*A_REF

                ## plot interlayer distance
                # if wanted: add auxiliary points to get consistent (fixed) scale of colorbar (there might be a better way, but this works nicely as well)
                if FIXED_COLORBAR:
                    x_sc.append(-1000); y_sc.append(-1000); d_local_sc.append(D_MIN) # minimum value
                    x_sc.append(-1000); y_sc.append(-1000); d_local_sc.append(D_MAX) # maximum value
                # add scatterplot of interlayer distance
                im = ax.scatter(x_sc, y_sc, c=d_local_sc, s=SIZE_PT,cmap=colormap) # scatter data

                x_sc = np.asarray(x_sc)
                y_sc = np.asarray(y_sc)
                d_local_sc = np.asarray(d_local_sc)

                # args_min_d = np.where(d_local_sc==min(d_local_sc))[0]
                args_max_d = np.where(d_local_sc==max(d_local_sc))[0]

                xs_max_d = x_sc[args_max_d]
                ys_max_d = y_sc[args_max_d]

                ##Algorithm to pick node index
                i_s = np.where((0.4*(X_POS-X_NEG)<xs_max_d) & (xs_max_d<X_POS)
                               & (-Y_NEG<ys_max_d) & (ys_max_d<0.5*(Y_POS-Y_NEG)))[0]

                top_i_s = i_s[np.where(ys_max_d[i_s]==max(ys_max_d[i_s]))[0]]
                top_left_i_s = top_i_s[np.argmin(xs_max_d[top_i_s])]
                # print(top_left_i_s)
                i=top_left_i_s

                points = np.column_stack((xs_max_d,ys_max_d))
                distances = cdist(points,points)[i]
                nn1_dist = np.unique(np.round(distances,10))[1] ##Get first nearest neighbours of node
                nn1s = np.where(np.round(distances,10) == nn1_dist)[0]
                nn1_arg_1 = nn1s[np.where((xs_max_d[nn1s]<xs_max_d[i]) & (ys_max_d[nn1s]>ys_max_d[i]))[0][0]]
                nn1_arg_2 = nn1s[np.where((xs_max_d[nn1s]>xs_max_d[i]) & (ys_max_d[nn1s]>ys_max_d[i]))[0][0]]
                x_sol1,x_sol2,x_sol3 = xs_max_d[[i,nn1_arg_1,nn1_arg_2]]
                y_sol1,y_sol2,y_sol3 = ys_max_d[[i,nn1_arg_1,nn1_arg_2]]
                x_cent,y_cent = (x_sol1+x_sol2+x_sol3)/3,(y_sol1+y_sol2+y_sol3)/3
                plt.scatter([x_sol1,x_sol2,x_sol3,x_cent],[y_sol1,y_sol2,y_sol3,y_cent],c='k',s=1)


                if m1 == m2:
                    x_min,x_max = x_cent,x_sol3
                else:
                    x_min,x_max = x_cent,x_sol3-0.2*nn1_dist
                # x_min,x_max = x_cent,x_sol3-0.05nn1_dist
                y_min,y_max = y_cent+0.1*nn1_dist,y_sol3+0.1*nn1_dist
                x_23_right,y_23_right = get_max_ds(x_sc,y_sc,d_local_sc,x_min,x_max,
                                  y_min,y_max,True,spacing=int(spacing*(x_max-x_min)/nn1_dist))
                #x_min,x_max = x_sol2+0.05*nn1_dist,x_cent

                if m1 == m2:
                    x_min,x_max = x_sol2,x_cent
                else:
                    x_min,x_max = x_sol2+0.2*nn1_dist,x_cent 
                y_min,y_max = y_cent+0.1*nn1_dist,y_sol2
                x_23_left,y_23_left = get_max_ds(x_sc,y_sc,d_local_sc,x_min,x_max,
                                  y_min,y_max,True,spacing=int(spacing*(x_max-x_min)/nn1_dist))
                xdata = np.concatenate(([x_sol2],x_23_left,x_23_right,[x_sol3]))
                ydata = np.concatenate(([y_sol2],y_23_left,y_23_right,[y_sol3]))
                plt.scatter(xdata,ydata,s=0.01,c='w')

                poly = np.polyfit(xdata,ydata,deg=poly_deg)
                x_23 = np.linspace(x_sol2,x_sol3,1000)
                y_23 = np.polyval(poly,x_23)
                plt.plot(x_23,y_23,c='w',lw=0.5)
                # plt.plot([x_sol2,x_sol3],[y_sol2,y_sol3],lw=0.5,linestyle='--',c='w')
                x_closest_23,y_closest_23 = get_closest_point(x_cent,y_cent,x_23,y_23)
                plt.scatter(x_closest_23,y_closest_23,c='k',s=1)
                xtangent_23,ytangent_23,m_23,c_23 = get_tangent(x_cent,x_closest_23,y_cent,y_closest_23,0.3*nn1_dist)
                plt.plot(xtangent_23,ytangent_23,linestyle='--',lw=0.5,c='k')

                inverted_xs, inverted_ys = mirror_image(x_23, y_23, y1=y_23[-1],x1=x_cent)
                #plt.plot(mirrored_xs,mirrored_ys,lw=0.5,c='w')

                x_12,y_12 = rotate_points(inverted_xs,inverted_ys,-60,x_23[0],y_23[0])
                plt.plot(x_12,y_12,lw=0.5,c='w')
                x_closest_12,y_closest_12 = get_closest_point(x_cent,y_cent,x_12,y_12)
                plt.scatter(x_closest_12,y_closest_12,c='k',s=1)
                xtangent_12,ytangent_12,m_12,c_12 = get_tangent(x_cent,x_closest_12,y_cent,y_closest_12,0.2*nn1_dist)
                plt.plot(xtangent_12,ytangent_12,linestyle='--',lw=0.5,c='k')

                x_13,y_13 = rotate_points(inverted_xs,inverted_ys,60,x_23[-1],y_23[-1])
                plt.plot(x_13,y_13,lw=0.5,c='w')
                x_closest_13,y_closest_13 = get_closest_point(x_cent,y_cent,x_13,y_13)
                plt.scatter(x_closest_13,y_closest_13,c='k',s=1)
                xtangent_13,ytangent_13,m_13,c_13 = get_tangent(x_cent,x_closest_13,y_cent,y_closest_13,0.1*nn1_dist)
                plt.plot(xtangent_13,ytangent_13,linestyle='--',lw=0.5,c='k')

                area_reduced = abs(np.trapz(y_23,x=x_23)-y_sol2*(x_sol3-x_sol2))
                area_triangle = ((x_sol3-x_sol2)**2)*(3**0.5)/4
                area_domain_MM = area_triangle-3*area_reduced
                area_domain_2H = area_triangle+3*area_reduced
                domain_ratio = area_domain_MM/area_domain_2H
                print(area_reduced,area_triangle,2*area_triangle,area_domain_MM,area_domain_2H,domain_ratio)
                ratios[k] = domain_ratio

                ax.set_xlim(-X_NEG, X_POS) # set x limits
                ax.set_ylim(-Y_NEG, Y_POS) # set y limits
                # remove ticks for x and y axis
                ax.get_xaxis().set_ticks([]) # don't show x-axis ticks
                ax.get_yaxis().set_ticks([]) # don't show y-axis ticks
                # tight layout
                #plt.tight_layout() # avoid problems with cut off labels
                # save plot and finish
                plt.savefig(image_file, dpi=DPI)
                #plt.savefig(f'heatmaps/{m1}{x1}2_{m2}{x1}2_{stacking}_{ii}.png',dpi=300)
                plt.close()
            ratios_dict[f'{stacking}-{m1}{m2}{x1}'] = ratios
```

```python
### Works perfectly for homobilayers

from matplotlib.patches import Arc

colormap='YlGnBu_r'
stacking = 'AP'
spacing = 100
SIZE_PT = 4 
ratios_dict = {}

poly_degs = np.array([4,5])
spacings = np.array([100,200,300])

spacing_factors = np.linspace(0.07,0.2,10)

for stacking in ['AP']:
    if stacking =='P':
        mpairs = [('Mo','W')]
    else:
        mpairs = [('Mo','Mo'),('Mo','W')]
            
    for m1,m2 in mpairs:
            
        for x1 in ['S','Se']:
            
            if stacking=='AP': 
                if m1==m2 or x1=='Se':
                    iis=[10,12,15,16,18,20,22,25,27,30,40]
                else:
                    iis=[10,12,15,16,18,20,22,25,27,30]
            else:
                iis=[16,18,20,22,25,27,30]
            
            print(f'Starting calculations for {stacking}-{m1}{x1}2/{m2}{x1}2...')
            
            if m1!=m2:
                x_stops = np.array([0.15,0.175,0.2])
                ratios = np.zeros((len(iis),len(spacing_factors)*len(x_stops)*len(poly_degs)))
            else:
                x_stops = np.array([0.0])
                ratios = np.zeros((len(iis),len(spacing_factors)))
            
            
            for num_ii,ii in enumerate(iis):
                print(f'ii={ii}')
                A_ML = 0.5*(a_optb88[f'{m1}{x1}2']+a_optb88[f'{m2}{x1}2'])


                if m1!=m2 and ii>10 and x1=='S':
                    filename = f'data_files/{m1}{x1}2_{m2}{x1}2_{stacking}_{ii}_twirl.xyz'
                else:
                    filename = f'data_files/{m1}{x1}2_{m2}{x1}2_{stacking}_{ii}.xyz'

                atoms = read(filename)
                cell = atoms.cell
                if stacking == 'P':
                    twist_angle = twist_angles[ii]
                elif stacking =='AP':
                    twist_angle = 60-twist_angles[ii]
                coord_LL,coord_UL = extract_positions(atoms)
                x_UL, y_UL, z_UL = coord_UL[:,0], coord_UL[:,1], coord_UL[:,2]
                x_LL, y_LL, z_LL = coord_LL[:,0], coord_LL[:,1], coord_LL[:,2]

                # extend coordinates to avoid interpolation artifacts
                x_UL_ext, y_UL_ext, z_UL_ext = ExtendCoordinates(x_UL, y_UL, z_UL, cell)
                x_LL_ext, y_LL_ext, z_LL_ext = ExtendCoordinates(x_LL, y_LL, z_LL, cell)
                # calculate interlayer distance
                x_grid, y_grid, A_REF = MakeGrid(cell, twist_angle,A_ML) # create interpolation grid
                z_UL_int = DoInterpolation(x_grid, y_grid, x_UL_ext, y_UL_ext, z_UL_ext) # interpolate UL
                z_LL_int = DoInterpolation(x_grid, y_grid, x_LL_ext, y_LL_ext, z_LL_ext) # interpolate LL
                d_local = GetInterlayerDistance(z_UL_int, z_LL_int) # calculate interlayer distance
                D_MIN, D_MAX = min(d_local), max(d_local) # Ang, interlayer distance references (min & max) to have consistent colorbar
                #print(d_local)
                ### perform plotting (not separated into functions)

                ## prepare parameters: size plotting area
                # window of (x,y) coord to plot relative to origin [Ang]
                X_NEG, X_POS = A_REF, 2*A_REF
                Y_NEG, Y_POS = int(0.5*A_REF), int(1.5*A_REF)
                # get cell constants
                a = np.sqrt(cell[0][0]**2+cell[0][1]**2+cell[0][2]**2) # cell constant a
                b = np.sqrt(cell[1][0]**2+cell[1][1]**2+cell[1][2]**2) # cell constant b
                # find how much to extend in terms of supercells
                x_left = math.ceil((X_NEG + 0.5*a)/a)+2
                x_right = math.ceil(X_POS/a)+1
                y_bottom = math.ceil(Y_NEG/(SIN60*b))+1
                y_top = math.ceil(Y_POS/(SIN60*b))+1

                ## prepare interpolated data: extend within supercell of plotting area
                x_sc, y_sc, d_local_sc = [], [], [] # prepare lists
                for m in range(-x_left-1, x_right+1): # supercell in x-direction
                    for n in range(-y_bottom-1, y_top+1): # supercell in y-direction
                        for i in range(len(d_local)): # over all data points
                            x_sc.append(x_grid[i]+m*cell[0][0]+n*cell[1][0]) # x coordinates
                            y_sc.append(y_grid[i]+m*cell[0][1]+n*cell[1][1]) # y coordinates
                            d_local_sc.append(d_local[i]) # local interlayer distance

                x_sc = np.asarray(x_sc)
                y_sc = np.asarray(y_sc)
                d_local_sc = np.asarray(d_local_sc)

                # args_min_d = np.where(d_local_sc==min(d_local_sc))[0]
                args_max_d = np.where(d_local_sc==max(d_local_sc))[0]

                xs_max_d = x_sc[args_max_d]
                ys_max_d = y_sc[args_max_d]

                ##Algorithm to pick node index
                i_s = np.where((0.4*(X_POS-X_NEG)<xs_max_d) & (xs_max_d<X_POS)
                               & (-Y_NEG<ys_max_d) & (ys_max_d<0.5*(Y_POS-Y_NEG)))[0]

                top_i_s = i_s[np.where(ys_max_d[i_s]==max(ys_max_d[i_s]))[0]]
                top_left_i_s = top_i_s[np.argmin(xs_max_d[top_i_s])]
                # print(top_left_i_s)
                i=top_left_i_s

                points = np.column_stack((xs_max_d,ys_max_d))
                distances = cdist(points,points)[i]
                nn1_dist = np.unique(np.round(distances,10))[1] ##Get first nearest neighbours of node
                nn1s = np.where(np.round(distances,10) == nn1_dist)[0]
                nn1_arg_1 = nn1s[np.where((xs_max_d[nn1s]<xs_max_d[i]) & (ys_max_d[nn1s]>ys_max_d[i]))[0][0]]
                nn1_arg_2 = nn1s[np.where((xs_max_d[nn1s]>xs_max_d[i]) & (ys_max_d[nn1s]>ys_max_d[i]))[0][0]]
                x_sol1,x_sol2,x_sol3 = xs_max_d[[i,nn1_arg_1,nn1_arg_2]]
                y_sol1,y_sol2,y_sol3 = ys_max_d[[i,nn1_arg_1,nn1_arg_2]]
                x_cent,y_cent = (x_sol1+x_sol2+x_sol3)/3,(y_sol1+y_sol2+y_sol3)/3
                
                m=0
                
                if m1==m2 or ii<15:
                    x_stops = np.array([0.0])
                elif m1!=m2 and ii==40:
                    x_stops = np.array([0.2,0.225,0.25])
                else:
                    x_stops = np.array([0.15,0.175,0.2])   
                
                
                for j,spacing_factor in enumerate(spacing_factors):
                    
                    for k,x_stop in enumerate(x_stops):
                        
                        spacing = int(nn1_dist/spacing_factor)//2
                        
                        
                                        
                        x_min,x_max = x_cent,x_sol3-x_stop*nn1_dist

                        
                        y_min,y_max = y_cent+0.1*nn1_dist,y_sol3+0.1*nn1_dist
#                         x_23_right,y_23_right = get_max_ds(x_sc,y_sc,d_local_sc,x_min,x_max,
#                                           y_min,y_max,True,spacing=int(spacing*(x_max-x_min)/nn1_dist))
                        x_23_right,y_23_right = get_max_ds(x_sc,y_sc,d_local_sc,x_min,x_max,
                                          y_min,y_max,True,spacing=spacing)

                        x_min,x_max = x_sol2+x_stop*nn1_dist,x_cent
            
                        y_min,y_max = y_cent+0.1*nn1_dist,y_sol2
#                         x_23_left,y_23_left = get_max_ds(x_sc,y_sc,d_local_sc,x_min,x_max,
#                                           y_min,y_max,True,spacing=int(spacing*(x_max-x_min)/nn1_dist))
                        x_23_left,y_23_left = get_max_ds(x_sc,y_sc,d_local_sc,x_min,x_max,
                                          y_min,y_max,True,spacing=spacing)
                        
                        xdata = np.concatenate(([x_sol2],x_23_left,x_23_right,[x_sol3]))
                        ydata = np.concatenate(([y_sol2],y_23_left,y_23_right,[y_sol3]))
                        
                        if len(x_stops)==1:
                            
                            area_reduced = abs(np.trapz(ydata,x=xdata)-y_sol2*(x_sol3-x_sol2))
                            area_triangle = ((x_sol3-x_sol2)**2)*(3**0.5)/4
                            area_domain_MM = area_triangle-3*area_reduced
                            area_domain_2H = area_triangle+3*area_reduced
                            domain_ratio = area_domain_MM/area_domain_2H
                            #print(area_reduced,area_triangle,2*area_triangle,area_domain_MM,area_domain_2H,domain_ratio)
                            ratios[num_ii][m] = domain_ratio
                            print(f'spacing={spacing}, x_stop={x_stop}, ratio={domain_ratio}')
                            m+=1
                        else:
                            
                            for l,poly_deg in enumerate(poly_degs):
                                
                                weights = np.ones(len(xdata))
                                weights[0],weights[-1]=100,100
                                poly = np.polyfit(xdata,ydata,w=weights,deg=poly_deg)
                                x_23 = np.linspace(x_sol2,x_sol3,1000)
                                y_23 = np.polyval(poly,x_23)
                                area_reduced = abs(np.trapz(y_23,x=x_23)-y_sol2*(x_sol3-x_sol2))
                                area_triangle = ((x_sol3-x_sol2)**2)*(3**0.5)/4
                                area_domain_MM = area_triangle-3*area_reduced
                                area_domain_2H = area_triangle+3*area_reduced
                                domain_ratio = area_domain_MM/area_domain_2H
                                #print(area_reduced,area_triangle,2*area_triangle,area_domain_MM,area_domain_2H,domain_ratio)
                                ratios[num_ii][m] = domain_ratio
                                print(f'spacing={spacing}, x_stop={x_stop}, poly_deg={poly_deg}, ratio={domain_ratio}')
                                m+=1
            ratios_dict[f'{stacking}-{m1}{m2}{x1}'] = ratios.copy()
```

```python
def parse_textfile_to_dict(filename):
    data_dict = {}

    with open(filename, "r") as file:
        for line in file:
            # Split at the first occurrence of ":", then strip whitespace
            parts = line.split(":", 1)
            if len(parts) < 2:
                continue  # Skip malformed lines
            
            key = parts[0].strip()
            value = parts[1].split("#")[0].strip()  # Remove comments
            
            # Convert numeric values
            if value.replace('.', '', 1).isdigit():  # Check for floats/integers
                value = float(value) if '.' in value else int(value)
            elif value == "{}":  # Handle empty placeholders
                value = None
            
            data_dict[key] = value

    return data_dict
# Example Usage
# file_path = "Exp_data/DS1/info.txt"
# data = parse_textfile_to_dict(file_path)
# print(data)
```
